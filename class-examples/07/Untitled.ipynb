```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# Dataset
data = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5, 6],
    "result": [0, 0, 0, 1, 1, 1]
})

# Features and Labels
X = data[["hours"]]
y = data["result"]

# Create model
model = DecisionTreeClassifier(max_depth=2)

# Train
model.fit(X, y)

# Predict
print(model.predict([[9]]))
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

# Dataset
data = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5, 6],
    "score": [45, 55, 68, 78, 88, 95]
})

# Features and Labels
X = data[["hours"]]
y = data["score"]

# Create model
model = DecisionTreeRegressor(max_depth=2)

# Train
model.fit(X, y)

# Predict
print(model.predict([[5]]))
```
